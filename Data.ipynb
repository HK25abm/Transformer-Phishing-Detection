```python
import pandas as pd

# Enron
enron = pd.read_csv("emails.csv")

# CEAS
ceas = pd.read_csv("CEAS_08.csv")

# Modern phishing
phishing = pd.read_csv("phishing.csv")

# Mixed dataset
paper2 = pd.read_csv("Phishing_Email.csv")

# Optional
legit = pd.read_csv("legit.csv")
```

```python
enron = enron.rename(
    columns={
        "message":"text"
    }
)
```

```python
enron = enron[
    ["text"]
]
```

```python
enron = enron.dropna()

enron = enron.drop_duplicates()
```

```python
enron["label"] = 0

enron["source"] = "Enron"

enron["urls"] = 0
```

```python
SEED = 42

enron = enron.sample(
    n=10000,
    random_state=SEED
)
```

```python
ceas["text"] = (
    ceas["subject"].fillna("")
    + " "
    + ceas["body"].fillna("")
)
```

```python
ceas = ceas[
    ceas["label"] == 1
]
```

```python
ceas = ceas[
    [
        "text",
        "label",
        "urls"
    ]
]
```

```python
ceas["source"] = "CEAS"
```

```python
phishing["text"] = (
    phishing["subject"].fillna("")
    + " "
    + phishing["body"].fillna("")
)
```

```python
phishing["source"] = "Modern"
```

```python
paper2 = paper2.rename(
    columns={
        "Email Text":"text"
    }
)
```

```python
paper2["label"] = paper2["Email Type"].map({
    "Safe Email":0,
    "Phishing Email":1
})
```

```python
paper2 = paper2[
    paper2["label"] == 1
]
```

```python
paper2["urls"] = 0
```

```python
paper2 = paper2[
    [
        "text",
        "label",
        "urls"
    ]
]
```

```python
paper2["source"] = "Paper2"
```

```python
final_df = pd.concat(
    [
        enron,
        ceas,
        phishing,
        paper2
    ],
    ignore_index=True
)
```

```python
final_df = final_df.drop_duplicates(
    subset=["text"]
)

final_df = final_df.dropna(
    subset=["text"]
)
```

```python
print(final_df["label"].value_counts())

print(final_df["source"].value_counts())
```

```python
legitimate = final_df[
    final_df["label"] == 0
]

phishing = final_df[
    final_df["label"] == 1
]

phishing = phishing.sample(
    n=10000,
    random_state=42
)

final_df = pd.concat(
    [
        legitimate,
        phishing
    ]
)
```

```python
final_df = final_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
final_df["sample_id"] = [
    f"EMAIL_{i:06d}"
    for i in range(
        1,
        len(final_df)+1
    )
]
```

```python
from sklearn.model_selection import train_test_split

train_df, temp_df = train_test_split(
    final_df,
    test_size=0.30,
    stratify=final_df["label"],
    random_state=42
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label"],
    random_state=42
)
```

```python
train_df.to_csv(
    "train.csv",
    index=False
)

val_df.to_csv(
    "validation.csv",
    index=False
)

test_df.to_csv(
    "test.csv",
    index=False
)

final_df.to_csv(
    "Final_Phishing_Email_Dataset.csv",
    index=False
)
```
