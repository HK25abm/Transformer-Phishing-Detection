```python
import re

def clean_email_text(text):
    text = str(text)

    parts = re.split(r"\n\s*\n", text, maxsplit=1)

    if len(parts) == 2:
        header, body = parts

        subject_match = re.search(
            r"(?im)^subject:\s*(.*)$",
            header
        )

        subject = (
            subject_match.group(1).strip()
            if subject_match
            else ""
        )

        text = f"{subject} {body}"

    text = re.sub(
        r"(?im)^(message-id|from|to|date|mime-version|content-type|content-transfer-encoding):.*$",
        " ",
        text
    )

    text = re.sub(r"<[^>]+>", " ", text)
    text = re.sub(r"[\r\n\t]+", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
def clean_frame(frame):
    frame = frame.copy()

    # Remove missing text or labels
    frame = frame.dropna(subset=["text", "label"])

    # Clean text
    frame["text"] = frame["text"].astype(str).str.strip()

    # Remove very short emails
    frame = frame[frame["text"].str.len() >= 20]

    # Ensure labels are integers
    frame["label"] = frame["label"].astype(int)

    # Keep only binary labels
    frame = frame[frame["label"].isin([0, 1])]

    # Remove duplicate emails
    frame = frame.drop_duplicates(subset=["text"])

    frame = frame.reset_index(drop=True)

    return frame
```

```python
enron = clean_frame(enron)
paper2 = clean_frame(paper2)
ceas = clean_frame(ceas)
modern = clean_frame(modern)

print("Enron :", len(enron))
print("Paper2:", len(paper2))
print("CEAS  :", len(ceas))
print("Modern:", len(modern))
```

```python
paper2_safe = paper2[
    paper2["label"] == 0
].copy()

paper2_phishing = paper2[
    paper2["label"] == 1
].copy()

print("Paper2 safe:", len(paper2_safe))
print("Paper2 phishing:", len(paper2_phishing))
```

```python
SEED = 42

enron_sample = enron.sample(
    n=8000,
    random_state=SEED
)

paper2_safe_sample = paper2_safe.sample(
    n=2000,
    random_state=SEED
)

ceas_sample = ceas.sample(
    n=6000,
    random_state=SEED
)

modern_sample = modern.sample(
    n=len(modern),
    random_state=SEED
)

paper2_phishing_needed = (
    10000
    - len(ceas_sample)
    - len(modern_sample)
)

paper2_phishing_sample = paper2_phishing.sample(
    n=paper2_phishing_needed,
    random_state=SEED
)

print("Paper2 phishing selected:", len(paper2_phishing_sample))
```

```python
master_df = pd.concat(
    [
        enron_sample,
        paper2_safe_sample,
        ceas_sample,
        paper2_phishing_sample,
        modern_sample
    ],
    ignore_index=True
)
```

```python
print(master_df.shape)

print(
    pd.crosstab(
        master_df["source"],
        master_df["label"],
        margins=True
    )
)
```

```python
before = len(master_df)

master_df = master_df.drop_duplicates(
    subset=["text"]
).reset_index(drop=True)

after = len(master_df)

print("Duplicates removed:", before - after)
print(master_df["label"].value_counts())
```

```python
legitimate_df = master_df[
    master_df["label"] == 0
]

phishing_df = master_df[
    master_df["label"] == 1
]

target_count = min(
    len(legitimate_df),
    len(phishing_df)
)

legitimate_df = legitimate_df.sample(
    n=target_count,
    random_state=SEED
)

phishing_df = phishing_df.sample(
    n=target_count,
    random_state=SEED
)

master_df = pd.concat(
    [legitimate_df, phishing_df],
    ignore_index=True
)

master_df = master_df.sample(
    frac=1,
    random_state=SEED
).reset_index(drop=True)
```

```python
master_df["stratify_key"] = (
    master_df["source"].astype(str)
    + "_"
    + master_df["label"].astype(str)
)
```

```python
from sklearn.model_selection import train_test_split

train_df, temp_df = train_test_split(
    master_df,
    test_size=0.30,
    random_state=SEED,
    stratify=master_df["stratify_key"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=SEED,
    stratify=temp_df["stratify_key"]
)
```

```python
for frame in [train_df, val_df, test_df]:
    frame.drop(
        columns=["stratify_key"],
        inplace=True
    )

    frame.reset_index(
        drop=True,
        inplace=True
    )
```
